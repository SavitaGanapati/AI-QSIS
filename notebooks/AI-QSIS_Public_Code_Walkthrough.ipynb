## 2. Synthetic example data

The rows and scores below are fictional examples for demonstrating table handling. They are not current market data and do not represent AI-QSIS production scores.

```python
import pandas as pd

example = pd.DataFrame([
    {"Stock": "EXAMPLE_A", "Sector": "Industrials", "FuturePotentialScore": 82.0, "RiskLevel": "MEDIUM"},
    {"Stock": "EXAMPLE_B", "Sector": "Energy", "FuturePotentialScore": 76.5, "RiskLevel": "LOW"},
    {"Stock": "EXAMPLE_C", "Sector": "Consumer", "FuturePotentialScore": 68.0, "RiskLevel": "HIGH"},
])

example
```

## 3. Basic data-quality checks

Before analysing a table, check expected fields, required values and numeric types. This is a generic example; production validation may be different or more extensive.

```python
required_columns = {"Stock", "Sector", "FuturePotentialScore", "RiskLevel"}
missing_columns = required_columns.difference(example.columns)

if missing_columns:
    raise ValueError(f"Missing required columns: {sorted(missing_columns)}")

if example["Stock"].isna().any():
    raise ValueError("Stock identifiers must not be missing in this example")

example["FuturePotentialScore"] = pd.to_numeric(
    example["FuturePotentialScore"], errors="coerce"
)
print("Rows:", len(example))
print("Missing values by column:")
print(example.isna().sum())
```

## 4. Sorting existing scores for review

This sorts fictional, already-computed values. It does **not** show how AI-QSIS calculates `FuturePotentialScore`, nor does it imply that a higher score predicts better investment performance.

```python
review_table = example.sort_values(
    by="FuturePotentialScore", ascending=False
).reset_index(drop=True)
review_table.insert(0, "ExampleRank", range(1, len(review_table) + 1))
review_table
```

## 5. Exporting a review table to Excel

This generic reporting pattern uses synthetic data. It is not a copy of the production workbook-generation code.

```python
from pathlib import Path

output_dir = Path("public_example_output")
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "synthetic_review_example.xlsx"

with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
    review_table.to_excel(writer, sheet_name="Example Ranking", index=False)

print(f"Example workbook created: {output_path}")
```
